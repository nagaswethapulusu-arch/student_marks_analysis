# Student Marks Analysis
This notebook calculates the total, average, highest and lowest marks.

```python
students = ["Asha", "Ravi", "Meena", "Kiran", "Swetha", "Harsh", "Rupesh", "Akhil"]
marks = [78, 85, 62, 90, 71, 89, 70, 65]
```

```python
total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)
```

```python
print("Total marks:", total)
print("Average marks:", round(average, 2))
print("Highest marks:", highest)
print("Lowest marks:", lowest)
```

```python
top = students[marks.index(highest)]
bottom = students[marks.index(lowest)]
print(f"Topper: {top} ({highest})")
print(f"Lowest: {bottom} ({lowest})")
```
